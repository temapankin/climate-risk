```python
import requests
import geopandas as gpd
import pandas as pd
import json
from pathlib import Path

BASE = "https://njmaps1.rad.rutgers.edu/arcgis/rest/services/CoastalFlooding/FEMA_NFHL_Effective_041220/MapServer/1"
OUT = Path("data/nj_nfhl_effective.parquet")

# Step 1: get all objectIds
r = requests.get(f"{BASE}/query", params={"where": "1=1", "returnIdsOnly": "true", "f": "json"}, timeout=60)
r.raise_for_status()
all_ids = r.json()["objectIds"]
print(f"Total features: {len(all_ids)}")

# Step 2: paginate in batches of 2000
BATCH = 2000
gdfs = []
for i in range(0, len(all_ids), BATCH):
    batch_ids = ",".join(map(str, all_ids[i : i + BATCH]))
    resp = requests.get(
        f"{BASE}/query",
        params={"objectIds": batch_ids, "outFields": "*", "returnGeometry": "true", "f": "geojson"},
        timeout=120,
    )
    resp.raise_for_status()
    gdf = gpd.read_file(resp.content, driver="GeoJSON")
    gdfs.append(gdf)
    print(f"  fetched {i + len(gdf)}/{len(all_ids)}", end="\r")

print()
result = pd.concat(gdfs, ignore_index=True)
result = gpd.GeoDataFrame(result, geometry="geometry", crs="EPSG:4326")
print(f"Combined: {len(result)} rows, CRS={result.crs}")

# Step 3: save
OUT.parent.mkdir(exist_ok=True)
result.to_parquet(OUT)
print(f"Saved → {OUT}")
result.head(3)
```
